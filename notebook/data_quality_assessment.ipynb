```python
# Cell 1: Import libraries and setup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')


plt.style.use('default')
sns.set_palette("husl")
%matplotlib inline

print(" FLIGHT PRICE DATA QUALITY ASSESSMENT")
```

```python
# Cell 2: Load the dataset with errors
try:
    df = pd.read_csv("../dataset/raw_data_with_errors.csv")
    print(" Dataset with errors loaded successfully!")
    print(f" Dataset shape: {df.shape}")
    print(f" Columns: {list(df.columns)}")
except FileNotFoundError:
    print(" File not found. Please run scripts/generate_data_errors.py first")
    raise

# Display basic info
print("\n BASIC DATASET INFO:")
print(df.info())
```

```python
# Cell 3: Comprehensive Data Overview
print("1.  COMPREHENSIVE DATA OVERVIEW")
print("-" * 50)

print("First 10 rows:")
display(df.head(10))

print("\nDescriptive statistics:")
display(df.describe(include='all'))

print("\nData types:")
print(df.dtypes)
```

```python
# Cell 4: Check for Missing Required Columns (Error Type 1)
print("2.  MISSING REQUIRED COLUMNS (Error Type 1)")


expected_columns = ['Airline', 'Source', 'Destination', 'Price', 'Duration', 'Date_of_Journey']
missing_columns = [col for col in expected_columns if col not in df.columns]

if missing_columns:
    print(" MISSING REQUIRED COLUMNS DETECTED:")
    for col in missing_columns:
        print(f"   - {col}")
    print(f"\n DEMO: Successfully detected {len(missing_columns)} missing required columns")
else:
    print(" All expected columns are present")

print(f"\nActual columns present: {list(df.columns)}")
```

```python
# Cell 5: Check for Missing Values (Error Type 2)
print("3.  MISSING VALUES ANALYSIS (Error Type 2)")


missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df)) * 100
missing_info = pd.DataFrame({
    'Missing Count': missing_data,
    'Missing Percentage': missing_percent
})
missing_info = missing_info[missing_info['Missing Count'] > 0]

if not missing_info.empty:
    print(" MISSING VALUES DETECTED:")
    display(missing_info)
    
    # Visualize missing values
    plt.figure(figsize=(12, 6))
    sns.heatmap(df.isnull(), cbar=True, yticklabels=False, cmap='viridis')
    plt.title('Missing Values Heatmap - Error Type 2')
    plt.tight_layout()
    plt.show()
    
    print(f" DEMO: Successfully detected missing values in {len(missing_info)} columns")
else:
    print(" No missing values found")
```

```python
# Cell 6: Check for Unknown Categories (Error Type 3)
print("4.  UNKNOWN CATEGORIES CHECK (Error Type 3)")


# Check Airline column for unknown values
if 'Airline' in df.columns:
    # Get unique airline values
    unique_airlines = df['Airline'].dropna().unique()
    print(f"Unique airlines found: {len(unique_airlines)}")
    print(f"All airlines: {list(unique_airlines)}")
    
    # Look for obviously fake airlines (from your error generation)
    suspicious_keywords = ['space', 'ocean', 'mountain']
    suspicious_airlines = []
    
    for airline in unique_airlines:
        if any(keyword in str(airline).lower() for keyword in suspicious_keywords):
            suspicious_airlines.append(airline)
    
    if suspicious_airlines:
        print(f"🚨 UNKNOWN/SUSPICIOUS AIRLINES DETECTED:")
        for airline in suspicious_airlines:
            print(f"   - {airline}")
        print(f" DEMO: Successfully detected {len(suspicious_airlines)} unknown category values")
    else:
        print(" No unknown airlines detected")
else:
    print("ℹ Airline column not available (was removed as part of Error Type 1)")
```

```python
# Cell 7: Check for Invalid Numeric Values (Error Type 4)
print("5.  INVALID NUMERIC VALUES (Error Type 4)")


# Check for negative Duration
if 'Duration' in df.columns:
    # Convert to numeric, handling errors
    df['Duration_numeric'] = pd.to_numeric(df['Duration'], errors='coerce')
    negative_duration = df[df['Duration_numeric'] < 0]
    
    if not negative_duration.empty:
        print(f" NEGATIVE DURATION VALUES DETECTED: {len(negative_duration)}")
        print("Sample negative durations:")
        display(negative_duration[['Duration']].head())
        
        # Visualize the issue
        plt.figure(figsize=(10, 6))
        valid_durations = df['Duration_numeric'][df['Duration_numeric'] >= 0].dropna()
        
        plt.subplot(1, 2, 1)
        plt.hist(valid_durations, bins=30, alpha=0.7, color='blue', edgecolor='black')
        plt.xlabel('Valid Duration')
        plt.ylabel('Frequency')
        plt.title('Valid Duration Distribution')
        
        plt.subplot(1, 2, 2)
        invalid_count = len(negative_duration)
        labels = ['Valid', 'Invalid (Negative)']
        sizes = [len(valid_durations), invalid_count]
        colors = ['lightblue', 'red']
        plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
        plt.title('Duration Data Quality')
        
        plt.tight_layout()
        plt.show()
        
        print(f" DEMO: Successfully detected {len(negative_duration)} invalid negative duration values")
    else:
        print(" No negative duration values found")
else:
    print("ℹ Duration column not available for checking")
```

```python
# Cell 8: Check for String Values in Numeric Columns (Error Type 5)
print("6.  STRING VALUES IN NUMERIC COLUMNS (Error Type 5)")


if 'Price' in df.columns:
    # Check for non-numeric values in Price
    price_numeric = pd.to_numeric(df['Price'], errors='coerce')
    string_prices = df[price_numeric.isna() & df['Price'].notna()]
    
    if not string_prices.empty:
        print(f" STRING VALUES IN PRICE COLUMN DETECTED: {len(string_prices)}")
        print("String values found in Price:")
        print(string_prices['Price'].unique())
        
        # Count of numeric vs string prices
        numeric_count = price_numeric.notna().sum()
        string_count = len(string_prices)
        
        # Visualization
        plt.figure(figsize=(12, 5))
        
        plt.subplot(1, 2, 1)
        labels = ['Numeric Values', 'String Values']
        sizes = [numeric_count, string_count]
        colors = ['lightgreen', 'red']
        plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
        plt.title('Price Column Data Types')
        
        plt.subplot(1, 2, 2)
        # Show distribution of numeric prices only
        numeric_prices_only = price_numeric.dropna()
        if len(numeric_prices_only) > 0:
            plt.hist(numeric_prices_only, bins=30, alpha=0.7, color='green', edgecolor='black')
            plt.xlabel('Price (numeric values only)')
            plt.ylabel('Frequency')
            plt.title('Numeric Price Distribution')
        
        plt.tight_layout()
        plt.show()
        
        print(f" DEMO: Successfully detected {string_count} string values in numeric Price column")
    else:
        print(" No string values found in Price column")
else:
    print("ℹ Price column not available for checking")
```

```python
# Cell 9: Check for Extreme Outliers (Error Type 6)
print("7.  EXTREME OUTLIERS DETECTION (Error Type 6)")

def detect_outliers_iqr(series):
    """Detect outliers using IQR method"""
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series[(series < lower_bound) | (series > upper_bound)]

# Check Duration for extreme outliers
if 'Duration' in df.columns:
    duration_numeric = pd.to_numeric(df['Duration'], errors='coerce').dropna()
    
    if len(duration_numeric) > 0:
        duration_outliers = detect_outliers_iqr(duration_numeric)
        extreme_outliers = duration_numeric[duration_numeric > 100]  # Your script sets outliers to 500
        
        print(f"Total duration outliers (IQR method): {len(duration_outliers)}")
        print(f"Extreme duration outliers (>100 hours): {len(extreme_outliers)}")
        
        if not extreme_outliers.empty:
            print(" EXTREME OUTLIERS DETECTED:")
            print(f"Extreme duration values: {extreme_outliers.unique()}")
            
            # Visualization
            plt.figure(figsize=(15, 5))
            
            plt.subplot(1, 3, 1)
            plt.boxplot(duration_numeric)
            plt.title('Duration - Box Plot')
            plt.ylabel('Duration (hours)')
            
            plt.subplot(1, 3, 2)
            plt.hist(duration_numeric, bins=30, alpha=0.7, color='orange', edgecolor='black')
            plt.axvline(x=100, color='red', linestyle='--', linewidth=2, label='Extreme threshold')
            plt.xlabel('Duration (hours)')
            plt.ylabel('Frequency')
            plt.title('Duration Distribution')
            plt.legend()
            
            plt.subplot(1, 3, 3)
            # Zoom in on normal range
            normal_durations = duration_numeric[duration_numeric <= 100]
            plt.hist(normal_durations, bins=20, alpha=0.7, color='lightblue', edgecolor='black')
            plt.xlabel('Duration (hours) - Normal Range')
            plt.ylabel('Frequency')
            plt.title('Normal Duration Distribution')
            
            plt.tight_layout()
            plt.show()
            
            print(f" DEMO: Successfully detected {len(extreme_outliers)} extreme outliers in Duration")
        else:
            print("No extreme outliers found in Duration")
    else:
        print("ℹ No valid duration data available for outlier detection")
else:
    print("ℹDuration column not available for checking")
```

```python
# Cell 10: Check for Inconsistent Date Formats (Error Type 7)
print("8.INCONSISTENT DATE FORMATS (Error Type 7)")


if 'Date_of_Journey' in df.columns:
    # Try to convert to datetime
    df['Date_parsed'] = pd.to_datetime(df['Date_of_Journey'], errors='coerce')
    invalid_dates = df[df['Date_parsed'].isna() & df['Date_of_Journey'].notna()]
    
    if not invalid_dates.empty:
        print(f" INVALID DATE FORMATS DETECTED: {len(invalid_dates)}")
        print("Invalid dates found:")
        invalid_date_examples = invalid_dates['Date_of_Journey'].unique()
        for date_example in invalid_date_examples:
            print(f"   - '{date_example}'")
        
        # Show valid vs invalid dates
        valid_dates_count = df['Date_parsed'].notna().sum()
        invalid_dates_count = len(invalid_dates)
        
        plt.figure(figsize=(10, 6))
        labels = ['Valid Dates', 'Invalid Dates']
        sizes = [valid_dates_count, invalid_dates_count]
        colors = ['lightgreen', 'red']
        
        plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
        plt.title('Date Format Validation')
        plt.show()
        
        print(f" DEMO: Successfully detected {invalid_dates_count} inconsistent date formats")
    else:
        print(" All dates have valid formats")
else:
    print("ℹ Date_of_Journey column not available for checking")
```

```python
# Cell 11: Comprehensive Data Quality Summary
print("9.COMPREHENSIVE DATA QUALITY SUMMARY")


print(" ALL ERROR TYPES DEMONSTRATION SUMMARY:")


error_detection_summary = []

# Error Type 1: Missing Columns
if missing_columns:
    error_detection_summary.append(f" Error Type 1: Missing Columns - {len(missing_columns)} columns missing")

# Error Type 2: Missing Values
if not missing_info.empty:
    total_missing = missing_info['Missing Count'].sum()
    error_detection_summary.append(f"Error Type 2: Missing Values - {total_missing} total missing values")

# Error Type 3: Unknown Categories
if 'Airline' in df.columns and 'suspicious_airlines' in locals() and suspicious_airlines:
    error_detection_summary.append(f" Error Type 3: Unknown Categories - {len(suspicious_airlines)} unknown airlines")

# Error Type 4: Invalid Numeric Values
if 'Duration' in df.columns and 'negative_duration' in locals() and not negative_duration.empty:
    error_detection_summary.append(f" Error Type 4: Invalid Numeric Values - {len(negative_duration)} negative durations")

# Error Type 5: String in Numeric Columns
if 'Price' in df.columns and 'string_prices' in locals() and not string_prices.empty:
    error_detection_summary.append(f" Error Type 5: String in Numeric - {len(string_prices)} string prices")

# Error Type 6: Extreme Outliers
if 'Duration' in df.columns and 'extreme_outliers' in locals() and not extreme_outliers.empty:
    error_detection_summary.append(f" Error Type 6: Extreme Outliers - {len(extreme_outliers)} extreme values")

# Error Type 7: Inconsistent Date Formats
if 'Date_of_Journey' in df.columns and 'invalid_dates' in locals() and not invalid_dates.empty:
    error_detection_summary.append(f" Error Type 7: Inconsistent Dates - {len(invalid_dates)} invalid dates")

# Display summary
for summary in error_detection_summary:
    print(f" {summary}")


print(f" DEMO READY: {len(error_detection_summary)}/7 error types successfully generated and detected!")
print(" You can now showcase each error type in your follow-up session!")

# Final dataset quality score
quality_score = len(error_detection_summary)  # Each detected error type = 1 point
max_score = 7
quality_percentage = (quality_score / max_score) * 100

print(f"\ DATA QUALITY SCORE: {quality_score}/{max_score} ({quality_percentage:.1f}%)")
print(" All generated data errors have been successfully identified!")
```
